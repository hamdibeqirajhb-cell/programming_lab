### **1. Trasformare cicli in list comprehension**

Fai tre esempi di cicli producono una lista e poi riscrivi lo stesso codice usando la list comprehension

```python
lista = []

for num in range(10):
    lista.append(num)

print(lista)

lc_lista = [num for num in range(10)]
print(lc_lista)


lista2 = []

for num in range(10):
    lista2.append(num*num)

print(lista2)

lc_lista2 = [num*num for num in range(10)]

print(lc_lista2)


lista3 = []

for num in range(10):
    for num2 in range(10,20):
        lista3.append(num + num2)

lc_lista3 = [num + num2 for num in range(10) for num2 in range(10,20)]

print(lista3)
print(lc_lista3)
```

### **2. Vettore di Numeri Primi**

1. **Crea un vettore** contenente tutti i numeri primi compresi tra **0 e 10**  
   _(Puoi scriverli direttamente nell'array)._

2. **Conta quanti numeri ci sono nel vettore** utilizzando la funzione `len()`.  **Ottieni lo stesso numero** accedendo all'attributo `.size` del vettore.

3. Quale pensi sia il tipo di dato (`dtype`) del vettore?  Prova a rispondere senza eseguire il codice e Verifica la tua risposta accedendo all'attributo `.dtype` del vettore

4. Scrivi l'array usando una list comprehension che controlla che i numeri siano primi. 

```python
import numpy as np

def is_prime(val):

    for i in range(2, val-1):
        if val % i == 0:
            return False
    return True

primi_10 = np.array([i for i in range(10) if is_prime(i)])

print("Primi 10 numeri primi: ", primi_10)
print("Lunghezza vettore: ", primi_10.size)
print("Lunghezza vettore: ", len(primi_10))

#il dtype dovrebbe essere array
print(primi_10.dtype)   #int64


    
```
