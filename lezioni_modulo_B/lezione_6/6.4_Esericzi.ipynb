# 4. Rispondere alle domande
Trasforma le seguenti richiesti in formule per estrarre infromazioni dal dataframe
- Mostrami le vendite maggiori di 10

- Mostrami i dati del 2018

- Mostrami le vendite maggiori di 13 e l'anno è il 2018

- Mostrami tutto TRANNE i casi in cui le vendite sono maggiori di 13 e l'anno è il 2018

- Mostrami i dati dove le vendite divise per 3 sono maggiori di 3

- Mostrami i dipendenti i cui nomi sono alfabeticamente dopo la J



```python
import pandas as pd

sales = pd.DataFrame(
    data={
        "employee": [
            "Katrina",
            "Guanyu",
            "Jan",
            "Roman",
            "Jacqueline",
            "Paola",
            "Esperanza",
            "Alaina",
            "Egweyn",
        ],
        "sales": [14, 17, 6, 12, 8, 3, 7, 15, 5],
        "year": [2018, 2019, 2020, 2018, 2020, 2019, 2019, 2020, 2020],
    }
)

#1 sales > 10
print(sales[sales["sales"] > 10])
print()

#2 year 2018
print(sales[sales["year"] == 2018])
print()

#3 sales > 13 e year == 2018
print(sales[(sales["sales"] > 13) & (sales["year"] == 2018)])
print()

#4 tutto tranne sales > 13 e l'anno è il 2018  --> tutti tranne gli elementi di #3
print(sales[(sales["sales"] < 13) | (sales["year"] != 2018)])
print()

#5 dati dove (sales / 3)   > 3
print(sales[(sales["sales"]/3) > 3])
print()

#6 dipendenti i cui nomi dopo la J
dipendenti = sales["employee"].values

content = []
for nome in dipendenti:
    if nome[0] > "J":
        content.append(nome)

print(sales[sales["employee"].isin(content)])
print()

#6
print(sales[(sales["employee"].str[0]) > "J"])
print()

sales
```

```python
# Import dei pacchetti
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Caricamento del dataset
url = 'https://zenodo.org/record/5898311/files/vgsales.csv'
df = pd.read_csv(url)
print(df.head())
print()


#1
print("Totale giochi considerati", len(df["Name"].unique()))

#1
print("Totale giochi considerati", df["Name"].nunique())
print()

#2 generi più popolari   --> barplot
"""plt.bar(data = df,x="Genre",y="Global_Sales")
plt.xlabel("Genere")
plt.ylabel("Vendite globali")
plt.title("Vendite per genere")
plt.show()
"""

#2 corretto
sale_gen = df.groupby("Genre")["Global_Sales"].sum().sort_values(ascending=False)
print(sale_gen)
plt.bar(sale_gen.index, sale_gen)
plt.xlabel("Genere")
plt.ylabel("Vendite globali")
plt.title("Vendite per genere")

plt.xticks(rotation = 90)
plt.show()

#2 seaborn
sale_gen2 = df.groupby("Genre", as_index=False)["Global_Sales"].sum().sort_values("Global_Sales", ascending=False)
print(sale_gen2)
sns.barplot(data=sale_gen2, 
            x="Genre",
            y="Global_Sales")
plt.title("Vendite per genere")
plt.xticks(rotation = 90)
plt.show()

#3 numero di giochi pubblicati nel tempo
plt.figure(figsize=(10,5))
g_nome = df.groupby("Year", as_index=False)["Name"].count().sort_values("Year")
#print("quiiii", type(g_nome["Year"].iloc[2]))
g_nome["Year"] = g_nome["Year"].astype(int)
sns.barplot(data=g_nome,
            x="Year",
            y="Name")
plt.ylabel("n° giochi")
plt.xticks(rotation = 90)
plt.show()

#3
gg_nome = df.groupby("Year")["Name"].size().sort_index()
print(gg_nome)
plt.bar(gg_nome.index, gg_nome)
plt.title("Giochi per anno")
plt.xticks(rotation = 90)
plt.show()

#4
"""
sns.barplot(data=df,
            x = "Genre",
            y = "Global_Sales",
            hue = ["NA_Sales", "EU_Sales", "JP_Sales", "Other_Sales"],
            multiple = "stack")
plt.show()
"""
#4
s_gen = df.groupby("Genre")[["NA_Sales", "EU_Sales", "JP_Sales", "Other_Sales"]].sum()
print(s_gen)
s_gen.plot(kind = "bar", stacked=True)
plt.xlabel("Genere")
plt.ylabel("Global_sale")
plt.show()
```

```python
#Quali sono i generi più popolari? Mostralo con un bar plot
genre_counts = df['Genre'].value_counts()

plt.figure(figsize=(10, 5))
genre_counts.plot(kind='bar')
plt.title('Giochi per Genere')
plt.ylabel('Numero di giochi')
plt.xlabel('Genere')
plt.xticks(rotation=45)
plt.tight_layout()
```

```python
#Plotta l'evoluzione del numero di giochi pubblicati nel tempo
games_per_year = df['Year'].value_counts().sort_index()

plt.figure(figsize=(10, 5))
games_per_year.plot(kind='line', marker='o')
plt.title('Evoluzione dei giochi pubblicati nel tempo')
plt.ylabel('Numero di giochi')
plt.xlabel('Anno')
plt.grid(True)
plt.tight_layout()
```

```python
top_generi = genre_counts.nlargest(5).index
#df_top = df[df['Genre'].isin(top_generi)]
df_top = df[df['Genre'].apply(lambda g: g in top_generi)]
vendite_regioni = df_top.groupby('Genre')[['NA_Sales', 'EU_Sales', 'JP_Sales', 'Other_Sales']].sum()
vendite_regioni.plot(kind='bar', stacked=True, title='Vendite per regione per genere')
plt.ylabel('Milioni di unità')
plt.tight_layout()
plt.show()
```
